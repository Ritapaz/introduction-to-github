```python
import numpy as np
import matplotlib.pyplot as plt

# Preços e renda
px = 2
py = 1
renda = 100

# Parâmetros da função de utilidade Cobb-Douglas
a, b = 0.5, 0.5

# Cálculo do ponto ótimo (demanda de Marshall)
x_opt = (a * renda) / px
y_opt = (b * renda) / py

# Nível de utilidade ótimo
U_opt = (x_opt**a) * (y_opt**b)

# Gerar curva de indiferença ótima
x = np.linspace(1, renda/px, 200)
y_indif_opt = (U_opt / (x**a))**(1/b)

# Restrição orçamentária
x_budget = np.linspace(0, renda/px, 200)
y_budget = (renda - px * x_budget) / py

# Plot
plt.figure(figsize=(8, 6))
plt.plot(x_budget, y_budget, label='Restrição Orçamentária')
plt.plot(x, y_indif_opt, label='Curva de Indiferença Ótima')
plt.scatter(x_opt, y_opt, color='red', label='Ponto Ótimo', zorder=5)
plt.xlabel('Bem X')
plt.ylabel('Bem Y')
plt.title('Ponto Ótimo do Consumidor (Demanda de Marshall)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Função para simular o comportamento do consumidor com sliders
def comportamento_consumidor(px=2, py=1, renda=100):
    a, b = 0.5, 0.5  # Parâmetros da função de utilidade Cobb-Douglas

    # Ponto ótimo (Demanda de Marshall)
    x_opt = (a * renda) / px
    y_opt = (b * renda) / py
    U_opt = (x_opt ** a) * (y_opt ** b)

    # Curva de indiferença ótima
    x = np.linspace(1, renda / px, 200)
    y_indif_opt = (U_opt / (x ** a)) ** (1 / b)

    # Restrição orçamentária
    x_budget = np.linspace(0, renda / px, 200)
    y_budget = (renda - px * x_budget) / py

    # Plot
    plt.figure(figsize=(8, 6))
    plt.plot(x_budget, y_budget, label='Restrição Orçamentária')
    plt.plot(x, y_indif_opt, label='Curva de Indiferença Ótima')
    plt.scatter(x_opt, y_opt, color='red', label='Ponto Ótimo', zorder=5)
    plt.xlabel('Bem X')
    plt.ylabel('Bem Y')
    plt.title('Escolha Ótima do Consumidor')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# Sliders para alterar preços e renda
interact(
    comportamento_consumidor,
    px=FloatSlider(value=2, min=0.5, max=10, step=0.5, description='Preço de X'),
    py=FloatSlider(value=1, min=0.5, max=10, step=0.5, description='Preço de Y'),
    renda=FloatSlider(value=100, min=20, max=300, step=10, description='Renda')
)

```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def simular_imposto(t=10):
    a, b = 100, 2   # Demanda: Qd = a - bP
    c, d = 20, 2    # Oferta:  Qs = -c + dP

    P = np.linspace(0, 80, 500)
    Qd = a - b * P
    Qs_sem_imposto = -c + d * P
    Qs_com_imposto = -c + d * (P - t)

    def encontrar_equilibrio(Qd, Qs, P):
        idx = np.argmin(np.abs(Qd - Qs))
        return P[idx], Qd[idx]

    P_eq, Q_eq = encontrar_equilibrio(Qd, Qs_sem_imposto, P)
    P_eq_imp, Q_eq_imp = encontrar_equilibrio(Qd, Qs_com_imposto, P)

    P_consumidor = P_eq_imp
    P_produtor = P_eq_imp - t
    receita_governo = t * Q_eq_imp

    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(Qd, P, label='Demanda', color='blue')
    ax.plot(Qs_sem_imposto, P, label='Oferta (sem imposto)', color='green')
    ax.plot(Qs_com_imposto, P, label='Oferta (com imposto)', color='red', linestyle='--')

    ax.fill_between([0, Q_eq_imp], P_consumidor, max(P), color='lightblue', alpha=0.4, label='Excedente do Consumidor')
    ax.fill_between([0, Q_eq_imp], min(P), P_produtor, color='lightgreen', alpha=0.4, label='Excedente do Produtor')
    ax.fill_between([0, Q_eq_imp], P_produtor, P_consumidor, color='orange', alpha=0.5, label='Receita do Governo')
    ax.fill_between([Q_eq_imp, Q_eq], P_produtor, P_consumidor, color='gray', alpha=0.4, label='Peso Morto')

    ax.scatter(Q_eq, P_eq, color='green', label='Equilíbrio sem imposto')
    ax.scatter(Q_eq_imp, P_eq_imp, color='red', label='Equilíbrio com imposto')

    ax.set_xlabel('Quantidade')
    ax.set_ylabel('Preço')
    ax.set_title(f'Simulação Interativa de Imposto (t = {t})\nReceita do Governo: {receita_governo:.2f}')
    ax.legend()
    ax.grid(True)
    plt.tight_layout()
    plt.show()

interact(
    simular_imposto,
    t=FloatSlider(value=10, min=0, max=30, step=1, description='Imposto t')
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros da demanda e oferta
a, b = 100, 2  # Demanda: Qd = a - bP
c, d = 20, 2   # Oferta: Qs = -c + dP
t = 10         # Imposto

# Vetor de preços
P = np.linspace(0, 80, 500)
Qd = a - b * P
Qs = -c + d * P
Qs_tax = -c + d * (P - t)

# Função para encontrar ponto de equilíbrio
def encontrar_equilibrio(Qd, Qs, P):
    idx = np.argmin(np.abs(Qd - Qs))
    return P[idx], Qd[idx]

# Equilíbrios
P_eq, Q_eq = encontrar_equilibrio(Qd, Qs, P)
P_eq_tax, Q_eq_tax = encontrar_equilibrio(Qd, Qs_tax, P)
P_produtor = P_eq_tax - t
P_consumidor = P_eq_tax

# Preço máximo e mínimo teóricos
P_max = a / b
P_min = c / d

# Gráfico
fig, ax = plt.subplots(figsize=(10, 6))

# Curvas
ax.plot(Qd, P, label='Demanda', color='blue')
ax.plot(Qs, P, label='Oferta (sem imposto)', color='green')
ax.plot(Qs_tax, P, label='Oferta (com imposto)', color='red', linestyle='--')

# Excedente do consumidor (abaixo da curva de demanda até o preço pago)
q_ec = np.linspace(0, Q_eq_tax, 200)
p_ec = (a - q_ec) / b
ax.fill_between(q_ec, p_ec, P_consumidor, where=p_ec > P_consumidor, color='lightblue', alpha=0.5, label='Excedente do Consumidor')

# Excedente do produtor (acima da curva de oferta até o preço recebido)
q_ep = np.linspace(0, Q_eq_tax, 200)
p_ep = (q_ep + c) / d
ax.fill_between(q_ep, p_ep, P_produtor, where=p_ep < P_produtor, color='lightgreen', alpha=0.5, label='Excedente do Produtor')

# Receita do governo (retângulo entre os preços)
ax.fill_between([0, Q_eq_tax], P_produtor, P_consumidor, color='orange', alpha=0.5, label='Receita do Governo')

# Peso morto (entre Q_tax e Q_eq)
q_pm = np.linspace(Q_eq_tax, Q_eq, 100)
p_dem_pm = (a - q_pm) / b
p_oferta_pm = (q_pm + c) / d
ax.fill_between(q_pm, p_oferta_pm, p_dem_pm, where=p_dem_pm > p_oferta_pm, color='gray', alpha=0.4, label='Peso Morto')

# Pontos de equilíbrio
ax.scatter(Q_eq, P_eq, color='green', label='Equilíbrio sem imposto')
ax.scatter(Q_eq_tax, P_eq_tax, color='red', label='Equilíbrio com imposto')

# Ajustes finais
ax.set_xlabel('Quantidade')
ax.set_ylabel('Preço')
ax.set_title("Simulação de Imposto: Áreas corretas dos excedentes e peso morto")
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.show()


interact(
    simular_imposto,
    t=FloatSlider(value=10, min=0, max=30, step=1, description='Imposto t')
)
```

```python
!pip install ipywidgets --quiet

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def simular_imposto_pigouviano(inter_cost=20.0, b_d=1.0, b_s=1.0):
    a_d = 100
    a_s = 20

    P = np.linspace(0, 150, 500)
    Q_d = a_d - b_d * P
    Q_s = -a_s + b_s * P
    Q_s_tax = -a_s + b_s * (P - inter_cost)

    def encontrar_equilibrio(Qd, Qs, P):
        idx = np.argmin(np.abs(Qd - Qs))
        return P[idx], Qd[idx]

    P_eq, Q_eq = encontrar_equilibrio(Q_d, Q_s, P)
    P_eq_tax, Q_eq_tax = encontrar_equilibrio(Q_d, Q_s_tax, P)

    P_produtor = P_eq_tax
    P_consumidor = P_eq_tax + inter_cost

    # Cálculo do bem-estar total
    def area_triangulo(base, altura):
        return 0.5 * base * altura

    ec_antes = area_triangulo(Q_eq, a_d - P_eq)
    ep_antes = area_triangulo(Q_eq, P_eq - (a_s / b_s))
    bem_estar_antes = ec_antes + ep_antes

    ec_depois = area_triangulo(Q_eq_tax, a_d - P_consumidor)
    ep_depois = area_triangulo(Q_eq_tax, P_produtor - (a_s / b_s))
    bem_estar_depois = ec_depois + ep_depois

    perda_total = bem_estar_antes - bem_estar_depois

    fig, ax = plt.subplots(figsize=(10, 6))

    ax.plot(Q_d, P, label="Demanda", color="blue")
    ax.plot(Q_s, P, label="Oferta (sem imposto)", color="green")
    ax.plot(Q_s_tax, P, label="Oferta (com imposto)", color="red", linestyle="--")

    ax.fill_between(np.linspace(0, Q_eq, 200), (a_d - np.linspace(0, Q_eq, 200)) / b_d, P_eq, color='lightblue', alpha=0.5, label="EC antes")
    ax.fill_between(np.linspace(0, Q_eq, 200), (np.linspace(0, Q_eq, 200) + a_s) / b_s, P_eq, color='lightgreen', alpha=0.5, label="EP antes")
    ax.fill_between(np.linspace(0, Q_eq_tax, 200), (a_d - np.linspace(0, Q_eq_tax, 200)) / b_d, P_consumidor, color='deepskyblue', alpha=0.5, label="EC após")
    ax.fill_between(np.linspace(0, Q_eq_tax, 200), (np.linspace(0, Q_eq_tax, 200) + a_s) / b_s, P_produtor, color='yellowgreen', alpha=0.5, label="EP após")

    ax.scatter(Q_eq, P_eq, color='green', label="Equilíbrio sem imposto")
    ax.scatter(Q_eq_tax, P_eq_tax, color='red', label="Equilíbrio com imposto")

    ax.set_xlabel("Quantidade")
    ax.set_ylabel("Preço")
    ax.set_title(
        f"Imposto Pigouviano = {inter_cost:.1f} | "
        f"EC+EP Antes = {bem_estar_antes:.1f}, Após = {bem_estar_depois:.1f}, "
        f"Δ Bem-estar = {perda_total:.1f}"
    )
    ax.legend()
    ax.grid(True)
    plt.tight_layout()
    plt.show()

interact(
    simular_imposto_pigouviano,
    inter_cost=FloatSlider(value=20, min=0, max=50, step=1, description='Imposto'),
    b_d=FloatSlider(value=1, min=0.2, max=5, step=0.2, description='Demanda'),
    b_s=FloatSlider(value=1, min=0.2, max=5, step=0.2, description='Oferta')
)
```

```python
!pip install ipywidgets --quiet

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def demanda(p, a=100, b=1):
    return np.maximum(0, a - b * p)

def oferta(p, c=20, d=1):
    return np.maximum(0, -c + d * p)

def simular_teto_preco(preco_maximo=60, b_d=1.0, d_o=1.0):
    a_d = 100  # intercepto da demanda
    c_o = 20   # intercepto da oferta

    p = np.linspace(0, 120, 500)
    q_d = demanda(p, a=a_d, b=b_d)
    q_s = oferta(p, c=c_o, d=d_o)

    # Equilíbrio sem teto
    def encontrar_equilibrio(Qd, Qs, P):
        idx = np.argmin(np.abs(Qd - Qs))
        return P[idx], Qd[idx]

    p_eq, q_eq = encontrar_equilibrio(q_d, q_s, p)

    # Com teto
    qd_teto = demanda(preco_maximo, a=a_d, b=b_d)
    qs_teto = oferta(preco_maximo, c=c_o, d=d_o)
    q_trocado = min(qd_teto, qs_teto)

    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(q_d, p, label='Demanda', color='blue')
    ax.plot(q_s, p, label='Oferta', color='green')

    ax.axhline(preco_maximo, color='red', linestyle='--', label=f'Teto de Preço = {preco_maximo}')
    ax.plot([qs_teto, qd_teto], [preco_maximo, preco_maximo], color='red', linewidth=3)
    ax.annotate('Excesso de Demanda',
                xy=((qs_teto + qd_teto) / 2, preco_maximo + 2),
                ha='center', color='red')

    # Excedente do consumidor
    ax.fill_between(np.linspace(0, q_trocado, 200),
                    preco_maximo,
                    (a_d - np.linspace(0, q_trocado, 200)) / b_d,
                    color='lightblue', alpha=0.5, label='Excedente do Consumidor')

    # Excedente do produtor
    ax.fill_between(np.linspace(0, q_trocado, 200),
                    (np.linspace(0, q_trocado, 200) + c_o) / d_o,
                    preco_maximo,
                    color='lightgreen', alpha=0.5, label='Excedente do Produtor')

    # Peso morto
    if q_trocado < q_eq:
        q_pm = np.linspace(q_trocado, q_eq, 100)
        p_d_pm = (a_d - q_pm) / b_d
        p_s_pm = (q_pm + c_o) / d_o
        ax.fill_between(q_pm, p_s_pm, p_d_pm,
                        color='gray', alpha=0.4, label='Peso Morto')

    ax.set_xlabel('Quantidade de Imóveis')
    ax.set_ylabel('Preço')
    ax.set_title('Teto de Preço: Excedentes e Peso Morto')
    ax.legend()
    ax.grid(True)
    plt.tight_layout()
    plt.show()

interact(
    simular_teto_preco,
    preco_maximo=FloatSlider(value=60, min=20, max=100, step=1, description='Teto de Preço'),
    b_d=FloatSlider(value=1.0, min=0.2, max=5.0, step=0.2, description='Elastic. Demanda'),
    d_o=FloatSlider(value=1.0, min=0.2, max=5.0, step=0.2, description='Elastic. Oferta')
)
```

```python
import random
import matplotlib.pyplot as plt

# Semente para resultados reproduzíveis
random.seed(42)

# Número de caminhões
num_caminhoes = 10

# Tempos entre chegadas (distribuição uniforme entre 5 e 15 minutos)
intervalos_chegada = [random.uniform(5, 15) for _ in range(num_caminhoes)]

# Tempo de chegada acumulado
tempos_chegada = [sum(intervalos_chegada[:i+1]) for i in range(num_caminhoes)]

# Tempo fixo de atendimento por caminhão
tempo_atendimento = 15

# Inicialização das listas de horários de início e fim do atendimento
inicios_atendimento = []
fins_atendimento = []

for i in range(num_caminhoes):
    if i == 0:
        inicio = tempos_chegada[i]
    else:
        # Começa a ser atendido quando chega ou quando o anterior termina, o que for mais tarde
        inicio = max(tempos_chegada[i], fins_atendimento[i-1])
    fim = inicio + tempo_atendimento

    inicios_atendimento.append(inicio)
    fins_atendimento.append(fim)

# Tempo de espera de cada caminhão
tempos_espera = [inicios_atendimento[i] - tempos_chegada[i] for i in range(num_caminhoes)]
tempos_totais = [fins_atendimento[i] - tempos_chegada[i] for i in range(num_caminhoes)]

# Gráfico do processo
plt.figure(figsize=(10, 6))
plt.plot(range(1, num_caminhoes + 1), tempos_chegada, marker='o', label="Chegada")
plt.plot(range(1, num_caminhoes + 1), inicios_atendimento, marker='^', label="Início do atendimento")
plt.plot(range(1, num_caminhoes + 1), fins_atendimento, marker='s', label="Término do atendimento")

plt.xlabel("Caminhão")
plt.ylabel("Tempo (minutos)")
plt.title("Simulação com Fila: Chegada e Atendimento de Caminhões")
plt.grid(True)
plt.legend()
plt.show()

# Estatísticas
tempo_medio_espera = sum(tempos_espera) / len(tempos_espera)
tempo_medio_total = sum(tempos_totais) / len(tempos_totais)

print(f"Tempo médio de espera: {tempo_medio_espera:.2f} minutos")
print(f"Tempo médio total no sistema: {tempo_medio_total:.2f} minutos")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider

def simular_fila_logistica(num_caminhoes=10, intervalo_min=5, intervalo_max=15, tempo_atendimento=15):
    np.random.seed(42)

    # Tempos entre chegadas aleatórios
    intervalos_chegada = np.random.uniform(intervalo_min, intervalo_max, size=num_caminhoes)
    tempos_chegada = np.cumsum(intervalos_chegada)

    inicios_atendimento = []
    fins_atendimento = []

    for i in range(num_caminhoes):
        if i == 0:
            inicio = tempos_chegada[i]
        else:
            inicio = max(tempos_chegada[i], fins_atendimento[i-1])
        fim = inicio + tempo_atendimento
        inicios_atendimento.append(inicio)
        fins_atendimento.append(fim)

    tempos_espera = [inicios_atendimento[i] - tempos_chegada[i] for i in range(num_caminhoes)]
    tempos_totais = [fins_atendimento[i] - tempos_chegada[i] for i in range(num_caminhoes)]

    # Gráfico
    plt.figure(figsize=(10, 6))
    plt.plot(range(1, num_caminhoes + 1), tempos_chegada, marker='o', label="Chegada")
    plt.plot(range(1, num_caminhoes + 1), inicios_atendimento, marker='^', label="Início Atendimento")
    plt.plot(range(1, num_caminhoes + 1), fins_atendimento, marker='s', label="Fim Atendimento")
    plt.xlabel("Caminhão")
    plt.ylabel("Tempo (minutos)")
    plt.title("Simulação com Fila: Chegada e Atendimento de Caminhões")
    plt.grid(True)
    plt.legend()
    plt.show()

    print(f"Tempo médio de espera: {np.mean(tempos_espera):.2f} minutos")
    print(f"Tempo médio total no sistema: {np.mean(tempos_totais):.2f} minutos")

interact(
    simular_fila_logistica,
    num_caminhoes=IntSlider(min=5, max=30, step=1, value=10, description='Caminhões'),
    intervalo_min=FloatSlider(min=1, max=10, step=0.5, value=5, description='Interv. Mín'),
    intervalo_max=FloatSlider(min=10, max=30, step=0.5, value=15, description='Interv. Máx'),
    tempo_atendimento=IntSlider(min=5, max=30, step=1, value=15, description='Atendimento')
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

def simular_estoque(dias=60, estoque_inicial=100, demanda_media=10, demanda_variacao=3,
                    ponto_reposicao=40, lote_pedido=80, lead_time=5, estoque_seguranca=0):

    np.random.seed(0)
    estoque = [estoque_inicial]
    pedidos_em_transito = []
    rupturas = 0

    estoque_dia = []
    dias_ruptura = []

    for dia in range(dias):
        # Verificar chegada de pedidos
        pedidos_chegando = [q for t, q in pedidos_em_transito if t == dia]
        if pedidos_chegando:
            estoque[-1] += sum(pedidos_chegando)
        pedidos_em_transito = [(t, q) for t, q in pedidos_em_transito if t > dia]

        # Demanda aleatória
        demanda = max(0, int(np.random.normal(demanda_media, demanda_variacao)))
        novo_estoque = estoque[-1] - demanda

        if novo_estoque < 0:
            rupturas += 1
            dias_ruptura.append(dia)
            novo_estoque = 0

        # Checar ponto de reposição
        if novo_estoque <= ponto_reposicao + estoque_seguranca:
            pedidos_em_transito.append((dia + lead_time, lote_pedido))

        estoque.append(novo_estoque)
        estoque_dia.append(novo_estoque)

    # Gráfico
    plt.figure(figsize=(12, 6))
    plt.plot(range(dias), estoque_dia, marker='o', label="Estoque")
    plt.axhline(ponto_reposicao, color='r', linestyle='--', label="Ponto de Reposição")
    plt.axhline(estoque_seguranca, color='orange', linestyle='--', label="Estoque de Segurança")
    if dias_ruptura:
        plt.scatter(dias_ruptura, [0]*len(dias_ruptura), color='black', label="Ruptura", zorder=5)
    plt.title("Simulação de Estoque com Política de Reposição")
    plt.xlabel("Dias")
    plt.ylabel("Nível de Estoque")
    plt.grid(True)
    plt.legend()
    plt.show()

    print(f"🟡 Média de estoque: {np.mean(estoque_dia):.2f}")
    print(f"🔴 Dias com ruptura de estoque: {rupturas} ({rupturas/dias:.1%})")

interact(
    simular_estoque,
    dias=IntSlider(min=30, max=120, step=10, value=60, description='Dias'),
    estoque_inicial=IntSlider(min=10, max=200, step=10, value=100, description='Estoque Inicial'),
    demanda_media=IntSlider(min=5, max=50, step=1, value=10, description='Demanda Média'),
    demanda_variacao=IntSlider(min=0, max=20, step=1, value=3, description='Variação'),
    ponto_reposicao=IntSlider(min=10, max=100, step=5, value=40, description='Ponto Rep.'),
    lote_pedido=IntSlider(min=20, max=200, step=10, value=80, description='Lote Pedido'),
    lead_time=IntSlider(min=1, max=15, step=1, value=5, description='Lead Time'),
    estoque_seguranca=IntSlider(min=0, max=50, step=5, value=0, description='Segurança')
)
```

```python
import simpy
import random
import matplotlib.pyplot as plt

TEMPO_SIMULACAO = 120  # minutos
TEMPO_ATENDIMENTO = 15  # minutos
CHEGADAS_POR_HORA = 4  # média de caminhões/hora

chegadas = []
esperas = []

def caminhão(env, nome, docas):
    chegada = env.now
    chegadas.append(chegada)
    print(f"{nome} chegou às {chegada:.2f}")

    with docas.request() as req:
        yield req
        espera = env.now - chegada
        esperas.append(espera)
        print(f"{nome} começou a carregar após {espera:.2f} minutos de espera")
        yield env.timeout(TEMPO_ATENDIMENTO)
        print(f"{nome} terminou às {env.now:.2f}")

def chegada_caminhoes(env, docas):
    i = 0
    while True:
        yield env.timeout(random.expovariate(CHEGADAS_POR_HORA / 60))
        i += 1
        env.process(caminhão(env, f'Caminhão {i}', docas))

env = simpy.Environment()
docas = simpy.Resource(env, capacity=2)
env.process(chegada_caminhoes(env, docas))
env.run(until=TEMPO_SIMULACAO)

# Análise gráfica
plt.hist(esperas, bins=10, color='skyblue', edgecolor='black')
plt.title("Distribuição dos Tempos de Espera")
plt.xlabel("Minutos")
plt.ylabel("Frequência")
plt.grid(True)
plt.show()

```

```python
import simpy
import random
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

def simular_fila(capacidade_docas, taxa_chegada, tempo_servico, tempo_total):
    random.seed(42)
    esperas = []

    def caminhão(env, nome, docas):
        chegada = env.now
        with docas.request() as req:
            yield req
            espera = env.now - chegada
            esperas.append(espera)
            yield env.timeout(tempo_servico)

    def chegada_caminhoes(env, docas):
        i = 0
        while True:
            yield env.timeout(random.expovariate(taxa_chegada / 60))
            i += 1
            env.process(caminhão(env, f'Caminhão {i}', docas))

    env = simpy.Environment()
    docas = simpy.Resource(env, capacity=capacidade_docas)
    env.process(chegada_caminhoes(env, docas))
    env.run(until=tempo_total)

    # Plot
    plt.figure(figsize=(10, 5))
    plt.hist(esperas, bins=15, color='lightblue', edgecolor='black')
    plt.title(f"Distribuição do Tempo de Espera\n{capacidade_docas} docas, chegada {taxa_chegada}/h, serviço {tempo_servico} min")
    plt.xlabel("Minutos de espera")
    plt.ylabel("Frequência")
    plt.grid(True)
    plt.show()

    print(f"Tempo médio de espera: {sum(esperas)/len(esperas):.2f} min")
    print(f"Número total de caminhões atendidos: {len(esperas)}")

interact(
    simular_fila,
    capacidade_docas=IntSlider(min=1, max=5, step=1, value=2, description='Docas'),
    taxa_chegada=IntSlider(min=2, max=20, step=1, value=5, description='Chegadas/h'),
    tempo_servico=IntSlider(min=5, max=30, step=1, value=15, description='Serviço (min)'),
    tempo_total=IntSlider(min=60, max=480, step=30, value=240, description='Tempo Sim. (min)')
)
```

```python
import streamlit as st
import simpy
import random
import matplotlib.pyplot as plt
import heapq

st.set_page_config(page_title="Simulação de Filas com Disciplinas", layout="centered")
st.title("🚚 Simulação de Filas com Diferentes Disciplinas")

st.markdown("""
Este aplicativo simula a operação de um centro de distribuição com diferentes **disciplinas de filas**:
- **FIFO**: ordem de chegada.
- **LIFO**: último a chegar é o primeiro a ser atendido.
- **Prioridade**: os caminhões com menor tempo de espera (mais urgentes) são priorizados.
""")

# Sidebar: parâmetros
st.sidebar.header("⚙️ Parâmetros")

docas = st.sidebar.slider("Número de Docas", 1, 5, 2)
taxa_chegada = st.sidebar.slider("Taxa de Chegada (caminhões/h)", 5, 30, 10)
tempo_servico = st.sidebar.slider("Tempo de Serviço (min)", 5, 30, 15)
tempo_total = st.sidebar.slider("Duração da Simulação (min)", 60, 480, 240)
disciplina = st.sidebar.selectbox("Disciplina da Fila", options=["FIFO", "LIFO", "Prioridade"])

esperas = []

# Simulação com diferentes disciplinas
def simular_fila():
    global esperas
    esperas = []
    random.seed(42)

    env = simpy.Environment()

    if disciplina == "FIFO":
        fila = simpy.Resource(env, capacity=docas)

        def caminhão(env, nome):
            chegada = env.now
            with fila.request() as req:
                yield req
                espera = env.now - chegada
                esperas.append(espera)
                yield env.timeout(tempo_servico)

        def chegada(env):
            i = 0
            while True:
                yield env.timeout(random.expovariate(taxa_chegada / 60))
                i += 1
                env.process(caminhão(env, f'Caminhão {i}'))

        env.process(chegada(env))

    elif disciplina == "LIFO":
        fila = []

        def doca(env):
            while True:
                if fila:
                    chegada, nome = fila.pop()  # Último a entrar, primeiro a sair
                    espera = env.now - chegada
                    esperas.append(espera)
                    yield env.timeout(tempo_servico)
                else:
                    yield env.timeout(1)

        def chegada(env):
            i = 0
            while True:
                yield env.timeout(random.expovariate(taxa_chegada / 60))
                i += 1
                fila.append((env.now, f'Caminhão {i}'))

        for _ in range(docas):
            env.process(doca(env))
        env.process(chegada(env))

    elif disciplina == "Prioridade":
        fila = []

        def doca(env):
            while True:
                if fila:
                    chegada, _, nome = heapq.heappop(fila)
                    espera = env.now - chegada
                    esperas.append(espera)
                    yield env.timeout(tempo_servico)
                else:
                    yield env.timeout(1)

        def chegada(env):
            i = 0
            while True:
                yield env.timeout(random.expovariate(taxa_chegada / 60))
                i += 1
                prioridade = random.randint(1, 100)
                heapq.heappush(fila, (env.now, prioridade, f'Caminhão {i}'))

        for _ in range(docas):
            env.process(doca(env))
        env.process(chegada(env))

    env.run(until=tempo_total)

# Executa a simulação
simular_fila()

# Resultados
st.subheader("📈 Resultados da Simulação")

if esperas:
    tempo_medio = sum(esperas) / len(esperas)
    st.metric("⏱️ Tempo Médio de Espera", f"{tempo_medio:.2f} min")
    st.metric("🚛 Caminhões Atendidos", len(esperas))

    fig, ax = plt.subplots()
    ax.hist(esperas, bins=15, color='lightgreen', edgecolor='black')
    ax.set_title("Distribuição dos Tempos de Espera")
    ax.set_xlabel("Minutos de espera")
    ax.set_ylabel("Frequência")
    st.pyplot(fig)
else:
    st.warning("Nenhum caminhão foi atendido na simulação.")
```

```python
import matplotlib.pyplot as plt
import networkx as nx
import itertools
import numpy as np

# Gerar pontos de entrega aleatórios
np.random.seed(0)
num_pontos = 6
pontos = np.random.rand(num_pontos, 2)

# Criar grafo completo com pesos de distância euclidiana
G = nx.complete_graph(num_pontos)
for i, j in G.edges():
    dist = np.linalg.norm(pontos[i] - pontos[j])
    G.edges[i, j]['weight'] = dist

# Encontrar melhor rota (força bruta para poucos pontos)
melhor_rota = None
menor_dist = float('inf')
for perm in itertools.permutations(range(num_pontos)):
    dist = sum(G.edges[perm[i], perm[i+1]]['weight'] for i in range(len(perm)-1))
    dist += G.edges[perm[-1], perm[0]]['weight']  # volta ao início
    if dist < menor_dist:
        menor_dist = dist
        melhor_rota = perm

# Visualizar a melhor rota
plt.figure(figsize=(8, 6))
for i, ponto in enumerate(pontos):
    plt.scatter(*ponto, color='blue')
    plt.text(ponto[0], ponto[1], f'{i}', fontsize=12)
rota = list(melhor_rota) + [melhor_rota[0]]
for i in range(len(rota)-1):
    p1, p2 = pontos[rota[i]], pontos[rota[i+1]]
    plt.plot([p1[0], p2[0]], [p1[1], p2[1]], 'r-')
plt.title(f"Melhor Rota Total: {menor_dist:.2f}")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx
import itertools
import numpy as np
from ipywidgets import interact, IntSlider

def roteirizacao_interativa(num_pontos=6, seed=0):
    if num_pontos > 9:
        print("Por favor, mantenha até 9 pontos para evitar lentidão.")
        return

    # Gerar pontos de entrega aleatórios
    np.random.seed(seed)
    pontos = np.random.rand(num_pontos, 2)

    # Criar grafo completo com pesos de distância euclidiana
    G = nx.complete_graph(num_pontos)
    for i, j in G.edges():
        dist = np.linalg.norm(pontos[i] - pontos[j])
        G.edges[i, j]['weight'] = dist

    # Encontrar melhor rota (força bruta)
    melhor_rota = None
    menor_dist = float('inf')
    for perm in itertools.permutations(range(num_pontos)):
        dist = sum(G.edges[perm[i], perm[i+1]]['weight'] for i in range(len(perm)-1))
        dist += G.edges[perm[-1], perm[0]]['weight']  # volta ao início
        if dist < menor_dist:
            menor_dist = dist
            melhor_rota = perm

    # Visualizar a melhor rota
    plt.figure(figsize=(8, 6))
    for i, ponto in enumerate(pontos):
        plt.scatter(*ponto, color='blue')
        plt.text(ponto[0], ponto[1], f'{i}', fontsize=12)
    rota = list(melhor_rota) + [melhor_rota[0]]
    for i in range(len(rota)-1):
        p1, p2 = pontos[rota[i]], pontos[rota[i+1]]
        plt.plot([p1[0], p2[0]], [p1[1], p2[1]], 'r-')
    plt.title(f"Melhor Rota Total: {menor_dist:.2f}")
    plt.grid(True)
    plt.show()

interact(
    roteirizacao_interativa,
    num_pontos=IntSlider(min=3, max=9, step=1, value=6, description='Nº Pontos'),
    seed=IntSlider(min=0, max=100, step=1, value=0, description='Seed Aleatória')
)
```
